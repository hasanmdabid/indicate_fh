```python
import numpy as np
from keras.models import load_model
from keras.utils import img_to_array, load_img
model_dir = '/home/abidhasan/Documents/Indicate_FH/saved_model/'
# Define paths to the models and images
model_paths = {
    'Vgg16': model_dir+'vgg16_100.h5',
    'Xception': model_dir+'xception_100.h5',
    'Inception': model_dir+'inceptionv3_100.h5',
    'Mobilenet': model_dir+'mobilenetv2_100.h5'
}

image_paths = {
    'img1': filtered_images_directory+'/original_0.png',
    'img2': filtered_images_directory+'/original_1.png',
    'img3': filtered_images_directory+'/original_2.png',
    'img4': filtered_images_directory+'/original_3.png',
    'img5': filtered_images_directory+'/original_4.png',
    'img6': filtered_images_directory+'/original_5.png',
    'img7': filtered_images_directory+'/original_6.png',
    'img8': filtered_images_directory+'/original_7.png',
    'img9': filtered_images_directory+'/original_8.png',
    'img10': filtered_images_directory+'/original_9.png',
    'img11': filtered_images_directory+'/original_10.png',
    'img12': filtered_images_directory+'/original_11.png',
    'img13': filtered_images_directory+'/original_12.png',
    'img14': filtered_images_directory+'/original_13.png',
    'img15': filtered_images_directory+'/original_14.png',
    
}

# Define the target size for each model (should match the input size expected by the model)
target_sizes = {
    'Vgg16': (224, 224),
    'Xception': (299, 299),
    'Inception': (299, 299),
    'Mobilenet': (224, 224)
}

# Load the models
models = {name: load_model(path) for name, path in model_paths.items()}

# Function to preprocess images


def preprocess_image(image_path, target_size):
    img = load_img(image_path, target_size=target_size)
    img_array = img_to_array(img)
    # Convert to a batch of one image
    img_array = np.expand_dims(img_array, axis=0)
    img_array /= 255.0  # Normalize to [0, 1]
    return img_array


# Perform predictions
for image_name, image_path in image_paths.items():
    print(f"\nPredictions for {image_name}:")
    for model_name, model in models.items():
        # Preprocess the image according to the model's expected input size
        img_array = preprocess_image(image_path, target_sizes[model_name])

        # Predict the class
        prediction = model.predict(img_array)

        # Since it's binary classification, the output is a single probability value
        # Convert probabilities to class labels (0 or 1)
        predicted_class = (prediction > 0.5).astype("int32")

        print(
            f"{model_name}: Predicted class: {predicted_class[0][0]}, Probability: {prediction[0][0]:.4f}")
```

```python
#Plotting the grad-cam effect of the shallow fine tuned effect (While the layers of the base models are Frozen) on hte CLE images,  

import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Number of images
num_images = 10

# Initialize lists to hold the images after reading them
original_images = []
filter1_images = []
filter2_images = []
filter3_images = []
filter4_images = []

# Read and store the images in their respective lists
for i in range(num_images):
    original_images.append(mpimg.imread(f'/home/abidhasan/Documents/Indicate_FH/grad_cam_images/original_{i}.png'))
    filter1_images.append(mpimg.imread(f'/home/abidhasan/Documents/Indicate_FH/grad_cam_images/xception_{i}_UF.png'))
    filter2_images.append(mpimg.imread(f'/home/abidhasan/Documents/Indicate_FH/grad_cam_images/mobilenetv2_{i}_UF.png'))
    filter3_images.append(mpimg.imread(f'/home/abidhasan/Documents/Indicate_FH/grad_cam_images/vgg16_{i}_UF.png'))
    filter4_images.append(mpimg.imread(f'/home/abidhasan/Documents/Indicate_FH/grad_cam_images/inceptionv3_{i}_UF.png'))

# Combine the images into a 2D list where each row contains the original and filtered versions
images = [original_images, filter1_images,
          filter2_images, filter3_images, filter4_images]

# Transpose the matrix
transposed_images = list(map(list, zip(*images)))

# Create a figure with a 10x5 grid of subplots (since it's transposed)
fig, axes = plt.subplots(nrows=10, ncols=5, figsize=(12, 18))

# Column titles
column_titles = ['Original', 'xception', 'mobilenetv2', 'vgg16', 'inceptionv3']

# Loop through the rows and columns to plot images
for i in range(10):
    for j in range(5):
        axes[i, j].imshow(transposed_images[i][j])
        axes[i, j].axis('off')  # Hide axes

# Add row numbers as labels on the leftmost side of each row
for i in range(10):
    axes[i, 0].text(-0.1, 0.5, str(i + 1), fontsize=12,
                    ha='center', va='center', transform=axes[i, 0].transAxes)

# Add column titles
for ax, col_title in zip(axes[0], column_titles):
    ax.set_title(col_title, fontsize=14, pad=10)

# Adjust layout to reduce gaps
plt.subplots_adjust(wspace=0.05, hspace=0.2)

# Show the plot
plt.tight_layout()
# Save the first plot
plt.savefig(filtered_images_directory+'/grad_cam_output_frozen.eps',
            format='eps', bbox_inches='tight')
```
